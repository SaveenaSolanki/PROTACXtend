```python
import matplotlib.pyplot as plt
from pathlib import Path

out = Path("original_quant_plot.png")

poses_sampled = 3600
poses_passed = 0

labels = ["Poses sampled", "Passed filter"]
values = [poses_sampled, poses_passed]

fig, ax = plt.subplots(figsize=(6,6), dpi=300)

bars = ax.bar(labels, values, edgecolor="black", linewidth=1.0)

ax.set_ylabel("Count")
ax.set_title("P4ward result: all sampled poses failed filtering")
ax.set_ylim(0, 4000)

for bar, val in zip(bars, values):
    ax.text(
        bar.get_x() + bar.get_width()/2,
        val + 60 if val > 0 else 60,
        f"{val}",
        ha="center",
        va="bottom",
        fontsize=12,
        fontweight="bold"
    )

annot = (
    "Closest gap = 10.83 Å\n"
    "Linker max = 0.74 Å\n"
    "~14.6× too short"
)

ax.text(
    0.98, 0.95, annot,
    transform=ax.transAxes,
    ha="right", va="top",
    bbox=dict(boxstyle="round,pad=0.3", facecolor="white", edgecolor="black")
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.savefig(out, dpi=300, bbox_inches="tight")
plt.show()
```
